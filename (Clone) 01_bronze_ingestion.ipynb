```python
# COMMAND -----------
import os

# Volume storage path inside Unity Catalog
STORAGE_ROOT = "/Volumes/dbacademy/ecommerce/medallion_files"

# Define path mapping for Medallion layers
PATHS = {
    "bronze": f"{STORAGE_ROOT}/bronze",
    "silver": f"{STORAGE_ROOT}/silver",
    "gold": f"{STORAGE_ROOT}/gold",
    "checkpoints": f"{STORAGE_ROOT}/_checkpoints"
}

# Create directories inside the Volume using standard os/dbutils
for layer, path in PATHS.items():
    dbutils.fs.mkdirs(path)
    print(f"Created/Verified path for [{layer.upper()}]: {path}")
```

```python
# COMMAND -----------
# Create dedicated database schemas for catalog navigation
spark.sql("CREATE DATABASE IF NOT EXISTS ecommerce_bronze;")
spark.sql("CREATE DATABASE IF NOT EXISTS ecommerce_silver;")
spark.sql("CREATE DATABASE IF NOT EXISTS ecommerce_gold;")

print("Medallion databases successfully created!")
```

```python
# COMMAND -----------
source_df = spark.table("dbacademy.default.`e-commerce_data`")

# Write to managed Unity Catalog table (UC manages the storage location)
(
    source_df.write
    .format("delta")
    .mode("overwrite")
    .saveAsTable("ecommerce_bronze.online_retail")
)

print("Bronze table successfully created: ecommerce_bronze.online_retail")
```
